```python
def get_mean_and_std(df, index):
    df_group = df.set_index(index)
    df_group = df.groupby(['label']).agg({index:['mean', 'std']}).reset_index()
    return df_group
```

```python
import seaborn as sns
```

```python
FONT_SIZE = 22
```

```python
font = {'family': 'verdana', 'size': FONT_SIZE}
plot_values = {'family': 'verdana', 'size': FONT_SIZE}
```

```python
def getLabelInMath(col_name):
    return r'$g_{' + str(col_name) + r'}$'
```

```python
a = sns.color_palette("Paired").as_hex()
b = sns.color_palette("hls", 8).as_hex()
red = b[0]
blue = a[1]
```

```python
def generate_boxplot(data, ax, labels, color, list_mean, gen):
    bplot = ax.boxplot(data, patch_artist=True, labels = labels)
    # ax.set_xlabel(getLabelInMath(gen + 1), fontname="verdana", fontsize=14)
    # ax.set_ylabel(getLabelInMath(gen + 1), fontname="verdana", fontsize=14)
    
    patch = bplot['boxes'][0]
    median = bplot['medians'][0]
    median_2 = bplot['medians'][1]
    median_2.set_color('black')
    median.set_color('black')
    patch.set_facecolor(color)
    ax.text(1.1, list_mean[0], round(list_mean[0], 2), fontweight='light', ha='left',va='bottom', fontdict=plot_values)
    ax.text(2.1, list_mean[1], round(list_mean[1], 2), fontweight='light', ha='left',va='bottom', fontdict=plot_values)
    # ax.set_title(f'Gen {gen + 1 }', fontdict=font)
    # ax.set_xticks([getLabelInMath(int(gen) + 1)])
    # title = ax.title(getLabelInMath(gen + 1))

    # Adjust the position of the title to the bottom
    # title.set_position([.5, 0])
    ax.tick_params(axis='x', labelsize=FONT_SIZE)
    ax.tick_params(axis='y', labelsize=FONT_SIZE)
    return ax
```

```python
def generate_bar_using_scalar(scalar, ax):
    ax.bar(range(1, len(scalar)+1), scalar)
    return ax
```

```python
def iterate_to_subset(array, df, labels):
    for gen in array:
        list_zero = df.loc[df['label'] == 0, str(gen)].to_numpy().tolist()
        list_one = df.loc[df['label'] == 1, str(gen)].to_numpy().tolist()
        list_wrap = [list_zero, list_one]
        mean_zero = np.mean(list_zero)
        mean_one = np.mean(list_one)
        list_mean = [mean_zero, mean_one]
        fig, ax = plt.subplots()
        ax.axhline(y=0, color='black', linestyle='--', linewidth=1, dashes=(5, 5))
        generate_boxplot(list_wrap, ax, labels, red, list_mean, gen)
        fig.savefig(f'figures/boxplot/gene_{gen + 1}.png',format='png', dpi =700)
        
```

```python
def set_box_color(bplot, color):
    # plt.setp(bp['boxes'], color=color)
    # plt.setp(bp['whiskers'], color=color)
    # plt.setp(bp['caps'], color=color)
    # plt.setp(bp['medians'], color='black')
    patch = bplot['boxes']
    for i in patch:
        i.set_facecolor(color)
```

```python
def set_medians_color(bplot):
    median = bplot['medians']
    for i in median:
        i.set_color('black')
```

```python
plt.rcParams.update(plt.rcParamsDefault)
plt.rcParams.update({
    'text.usetex' : False,
    "font.family": "Verdana",
    "font.monospace": 'Verdana'
})
```

```python
prognosis = ('Negative', 'Positive')
def generate_grouped_boxplot(subset):
    positive_cases = []
    negative_cases = []
    genes_ticks = [gen for gen in subset]
    for gen in subset:
        list_zero = df.loc[df['label'] == 0, int(gen)].to_numpy().tolist()
        list_one = df.loc[df['label'] == 1, int(gen)].to_numpy().tolist()
        negative_cases.append(list_zero)
        positive_cases.append(list_one)
    width = 8
    height = 6
    figsize=(width, height)
    fig, ax = plt.subplots(figsize = figsize)
    ax.axhline(y=0, color='black', linestyle='--', linewidth=1, dashes=(5, 5))
    bpl = ax.boxplot(negative_cases, positions=np.array(range(len(negative_cases)))*3.0-0.4, patch_artist=True)
    bpr = ax.boxplot(positive_cases, positions=np.array(range(len(positive_cases)))*3.0+0.4, patch_artist=True)
    set_box_color(bpl, red) 
    set_medians_color(bpr)
    set_medians_color(bpl)
    print('blue: ', blue)
    ax.plot([], c=red, label='Negative')
    ax.plot([], c=blue, label='Positive')
    ax.legend(prop=font)
    ticks_post = [i * 1.45 for i in list(range(0, len(genes_ticks) * 2, 2)) ]
    ax.set_xticks(ticks_post, [getLabelInMath(int(gen) + 1) for gen in genes_ticks])
    ax.tick_params(axis='x', labelsize=14)
    ax.tick_params(axis='y', labelsize=14)
    # ax.set_xlabel('Gen', fontname="verdana", fontsize=14)
    # ax.set_ylabel('Mean', fontname="verdana", fontsize=14)
    fig.show()
    fig.savefig(f'figures/boxplot/profile_10_gene.png',format='png', dpi =700)
```
