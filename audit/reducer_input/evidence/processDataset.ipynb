```python
def processDataSet(data_name):
    data = pd.read_csv('../data/' + data_name)
    labels = data['label'].to_numpy(dtype=np.intp)
    data = data.drop(['label'], axis= 1).to_numpy()
    return data, labels
```

```python
data, labels = processDataSet('rtfDataSet.csv')
data
```

```python
scaler = MinMaxScaler()
scaler.fit(data)
procesed = scaler.transform(data)
procesed
```

```python
complete = np.insert(procesed, procesed.shape[1], labels, 1)
complete
```

```python
complete.shape
```

```python
cols = [i for i in range(procesed.shape[1])]
cols.append('label')
dataNormal = pd.DataFrame(complete, columns= cols)
dataNormal['label'] = dataNormal['label'].astype(int)
dataNormal
```

```python
dataNormal.to_csv('data/Normal_All295_101POS_194NEG.csv', sep=',', index=False)
```
